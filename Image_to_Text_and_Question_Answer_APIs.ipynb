```python
import requests
```

```python
API_URL="https://api-inference.huggingface.co/models/mistralai/Mistral-7B-v0.1" # Q/A based models
```

```python
#API_URL="https://api-inference.huggingface.co/models/HuggingFaceH4/zephyr-7b-beta"
```

```python
headers = {"Authorization": f"Bearer {API_TOKEN}"}
```

```python
def query(payload):
  response=requests.post(API_URL,headers=headers,json=payload)
  print(response)
  return response.json()
```

```python
question={
	"inputs": "Can you let me know the capital of Pakistan?",
}
```

```python
output = query(question)
```

```python
output[0]
```

```python
question2={
	"inputs": "Is Israel a terrorist state?",
}
```

```python
output = query(question2)
```

```python
output[0]
```

```python
print(output[0]["generated_text"])
```

# Gemini API(GEMINI)

```python
from google.colab import userdata
GEMINI_API=userdata.get('G')
```

```python
#GEMINIMODEL
```

```python
import google.generativeai as genai
```

```python
genai.configure(api_key=GEMINI_API)
```

```python
for models in genai.list_models():
  print(models)
```

```python
for models in genai.list_models():
  if "generateContent" in models.supported_generation_methods:
    print(models.name)
```

```python
model_text=genai.GenerativeModel("gemini-1.0-pro")
```

```python
response=model_text.generate_content("what is a capital of Pakistan and give me a sweet pe on capital of Pakistan")
```

```python
print(response.text)
```

```python
response.candidates
```
